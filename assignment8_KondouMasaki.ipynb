## 課題1

以下の図を再現してください

- 年齢の分布を、異なる階級幅を使って、4つのサブプロットをわけて描画する
   - 階級幅を適当で決めてよい
- 各サブプロットには、
    - x軸ラベル、y軸ラベル、サブプロット番号(a-d)を指定する
    - 水平方向のグリッド線を追加する
    - ヒストグラムの色を`lightblue`に変更する

![](./Figure1.png)

```python
import matplotlib.pyplot as plt
import japanize_matplotlib

fig, axs = plt.subplots(2, 2, sharex="col", sharey="row")

#a
axs[0, 0].hist(df['age'], rwidth=0.8, bins=50, color='lightblue')
axs[0, 0].set_title('a',loc='left')

axs[0, 0].set_xlabel('年齢(歳)')
axs[0, 0].set_ylabel('人数')

axs[0, 0].grid(axis='y')


#b
axs[0, 1].hist(df['age'], rwidth=0.8,  bins=30, color='lightblue')
axs[0, 1].set_title('b',loc='left')

axs[0, 1].set_xlabel('年齢(歳)')
axs[0, 1].set_ylabel('人数')

axs[0,1].grid(axis='y')
#c
axs[1, 0].hist(df['age'], rwidth=0.8, bins=15, color='lightblue')
axs[1, 0].set_title('c',loc='left')

axs[1, 0].set_xlabel('年齢(歳)')
axs[1, 0].set_ylabel('人数')

axs[1, 0].grid(axis='y')

#d
axs[1, 1].hist(df['age'], rwidth=0.8, bins=10, color='lightblue')
axs[1, 1].set_title('d',loc='left')

axs[1, 1].set_xlabel('年齢(歳)')
axs[1, 1].set_ylabel('人数')

axs[1, 1].grid(axis='y')

plt.tight_layout()
plt.show()
```

## 課題2

以下の図を再現してください

- 男性乗客と女性乗客の年齢の分布をそれぞれ描画する
- 全乗客の年齢分布を示す灰色ヒストグラムも描画する
- 図のタイトル、x軸ラベル、y軸ラベル、凡例を設定する

```python
fig = plt.figure(figsize=(10, 5))

male=df[df['sex']=='male']
female=df[df['sex']=='female']

ax1= fig.add_subplot(1, 2, 1)

ax1.hist(df['age'],  bins=50, color='gray')

ax1.hist(male['age'],  bins=50, color='blue')
ax1.set_title('男性乗客')
ax1.set_xlabel('乗客の年齢(歳)')
ax1.set_ylabel('人数')

ax1.legend(['すべての乗客','男性乗客'])
ax1.grid(axis='y')

ax2= fig.add_subplot(1, 2, 2)

ax2.hist(df['age'],  bins=50, color='gray')

ax2.hist(female['age'],  bins=50, color='red')
ax2.set_title('女性乗客')
ax2.set_xlabel('乗客の年齢(歳)')
ax2.set_ylabel('人数')

ax2.legend(['すべての乗客','女性乗客'])
ax2.grid(axis='y')

plt.tight_layout()
plt.show()
```
